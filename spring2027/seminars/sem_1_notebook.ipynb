# 3. Операции с столбцами, функции groupby и agg (аналог оконных функций SQL)
## ❓ С кем нам удобнее работать - с пользователями или отдельными сессиями?<br> Сделаем новый столбец - дата регистрации

```python
# на каждого юзера в общем случае несколько сессий
# надо найти время самой первой сессии - новый столбик

df['registered_dttm'] = (
    df.groupby('user_id')['session_start_dttm']
      .transform('min')
)
```

```python
# проверим глазами, что столбец создался правильно

print(df.sort_values(by=['user_id', 'session_start_dttm']).head(20))
```

## ❓Как часто люди пользуются нашим продуктом и нравится ли он?<br> Сделаем новые столбцы - количество сессий и среднюю удовлетворенность

```python
sessions = df.groupby(['user_id']).agg({'session_start_dttm':'count'})
sessions = sessions.rename(columns={'session_start_dttm':'sessions_total'})
sessions
```

```python
satisfaction = df.groupby(['user_id']).agg({'satisfaction_metric':'mean'})
satisfaction = satisfaction.rename(columns={'satisfaction_metric':'satisfaction_avg'})
satisfaction
```

## ❓Какое будет среднее количество сессий на пользователя (почему мы не можем сказать, не запуская следующие ячейки)? <br>Какая будет средняя удовлетворенность (почему она не совпадет с 1.288082 из 5 ячейки)? 

```python
sessions.describe()
```

```python
satisfaction.describe()
```

```python
#добавим новые столбцы
df = pd.merge(df, sessions, how='left', right_on=['user_id'], left_on=['user_id'])
df_result = pd.merge(df, satisfaction, how='left', right_on=['user_id'], left_on=['user_id'])
df_result.sort_values(by=['user_id', 'session_start_dttm']).head(15)
```

## Выгрузка

```python
df_result.to_csv('sessions_1_result.csv', index=False)
```

## ❓Выгрузим топ-10 самых лояльных  и самых не лояльных пользователей (Зачем бизнесу их знать? Что с ними можно делать?) <br> 

```python
users_satisf=df_result[['user_id','sessions_total']].drop_duplicates(subset='user_id')

most_loyal=users_satisf.sort_values(by=['sessions_total']).head(10)
most_loyal
```

```python
least_loyal=users_satisf.sort_values(by=['sessions_total']).tail(10)
least_loyal
```

## ❓Зависит ли количество сессий от удовлетворенности? Посмотрим среднюю удовлетворенность по топ-10 и антитоп-10 <br> 

```python
most_loyal_info = pd.merge(df_result, most_loyal, how='right', right_on=['user_id'], left_on=['user_id'])
most_loyal_info
```

```python
most_loyal_info['satisfaction_avg'].mean()
```

```python
least_loyal_info = pd.merge(df_result, least_loyal, how='right', right_on=['user_id'], left_on=['user_id'])
least_loyal_info
```

```python
least_loyal_info['satisfaction_avg'].mean()
```

## ❓Есть ли корреляция между количеством сессий и удовлетворенностью от нашего продукта? Интуитивно - должна быть, хотя значения 1.206 и 1.303 очень близки 

## ❓Как правильно посчитать корреляцию, чтобы пользователи не задваивались из-за разных сессий?

```python
users_sessions_satisf=df_result[['user_id','sessions_total', 'satisfaction_avg']].drop_duplicates(subset='user_id') 
#оставим только уникальных пользователей
users_sessions_satisf['satisfaction_avg'].corr(users_sessions_satisf['sessions_total'])
```

```python
users_sessions_satisf['satisfaction_avg'].corr(users_sessions_satisf['sessions_total'], method='spearman') 
#более устойчивый к выбросам метод
```
